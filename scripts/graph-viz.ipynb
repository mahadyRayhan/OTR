```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- Journal Quality Settings (from user style) ---
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 20,
    'font.weight': 'bold',
    'axes.labelsize': 30, 
    'axes.titlesize': 34,
    'xtick.labelsize': 24,
    'ytick.labelsize': 24,
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
    'legend.fontsize': 24,
    'hatch.linewidth': 2.0 
})

sns.set_style("whitegrid", {'grid.linestyle': '--'})

# --- Data Extraction from Image ---
categories = ['Correctness', 'Completeness', 'Grounding', 'Specificity', 'Code Quality', 'Coherence']

# Subplot (a) Data
with_profiler = [4.60, 4.43, 2.40, 5.83, 3.20, 7.13]
no_profiler = [4.33, 4.50, 1.90, 5.40, 2.93, 6.97]

# Subplot (b) Data
deltas = [0.27, -0.07, 0.50, 0.43, 0.27, 0.16]

# --- Plotting ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(24, 11))
# fig.suptitle('Profiler Ablation (N = 30)', fontsize=40, fontweight='bold', y=1.05)

# ---------------------------------------------------
# Subplot (a): Per-Dimension Scores
# ---------------------------------------------------
bar_width = 0.35
r1 = np.arange(len(categories))
r2 = [x + bar_width for x in r1]

# Colors & Textures mapped to target image but using provided style
c1_face = '#41ab78' # Greenish 
c1_edge = '#206b47'
hatch1 = '//'

c2_face = '#a0a8b0' # Grayish 
c2_edge = '#5e656d'
hatch2 = '..'

bars1 = ax1.bar(r1, with_profiler, color=c1_face, width=bar_width, 
                edgecolor=c1_edge, linewidth=2, hatch=hatch1, 
                label='With Profiler')

bars2 = ax1.bar(r2, no_profiler, color=c2_face, width=bar_width, 
                edgecolor=c2_edge, linewidth=2, hatch=hatch2, 
                label='No Profiler')

# Value Labels adjusted for 0-10 scale
def add_labels_ax(ax, bars, vals):
    for bar, val in zip(bars, vals):
        height = bar.get_height()
        y_pos = height + 0.15 
        ax.text(
            bar.get_x() + bar.get_width() / 2, 
            y_pos,
            f'{val:.2f}', 
            ha='center', 
            va='bottom', 
            color='black', 
            fontsize=20, 
            fontweight='bold'
        )

add_labels_ax(ax1, bars1, with_profiler)
add_labels_ax(ax1, bars2, no_profiler)

# Formatting (a)
ax1.set_ylabel('Mean Judge Score (0–10)')
ax1.set_title('(a) Per-Dimension Scores', pad=20)
ax1.set_xticks([r + bar_width/2 for r in range(len(categories))])
ax1.set_xticklabels(categories, rotation=20, ha='right')
ax1.set_ylim(0, 9)
ax1.legend(loc='upper right', frameon=False)

# ---------------------------------------------------
# Subplot (b): Profiler Effect per Dimension
# ---------------------------------------------------
y_pos = np.arange(len(categories))

c_neg_face = '#e25c53' # Reddish
c_neg_edge = '#a82c24'

for i, delta in enumerate(deltas):
    color = c1_face if delta >= 0 else c_neg_face
    edge = c1_edge if delta >= 0 else c_neg_edge
    hatch = hatch1 if delta >= 0 else '\\\\'
    
    ax2.barh(i, delta, color=color, edgecolor=edge, linewidth=2, height=0.7, hatch=hatch)
    
    # Dynamic text positioning based on positive/negative
    align = 'left' if delta >= 0 else 'right'
    x_offset = 0.02 if delta >= 0 else -0.02
    ax2.text(delta + x_offset, i, f'{delta:+.2f}', va='center', ha=align, 
             fontsize=20, fontweight='bold', color='black')

# Formatting (b)
ax2.set_yticks(y_pos)
ax2.set_yticklabels(categories)
ax2.invert_yaxis() # Order Correctness at top
ax2.set_xlabel('Score Delta (With − No Profiler)')
ax2.set_title('(b) Profiler Effect per Dimension', pad=20)
ax2.set_xlim(-0.25, 0.7)
ax2.axvline(0, color='gray', linewidth=1.5, zorder=0)

sns.despine(ax=ax1)
sns.despine(ax=ax2)

plt.tight_layout()

# --- Save ---
plt.savefig('profiler_ablation_styled.png', dpi=600, bbox_inches='tight')
plt.savefig('profiler_ablation_styled.pdf', format='pdf', bbox_inches='tight')

print("Saved 'profiler_ablation_styled.png'")
# print("Saved 'profiler_ablation_styled.pdf'")

# Reset defaults
plt.rcParams.update(plt.rcParamsDefault)
```
